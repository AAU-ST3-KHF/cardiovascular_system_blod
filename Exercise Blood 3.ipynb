# Helper functions

```python
import numpy as np


# Poiseuille's Law: Q = (π * ΔP * r^4) / (8 * η * L)
def poiseuille_flow(dP, radius, viscosity, length):
    return (np.pi * mmHg2Pa(dP) * radius**4) / (8 * viscosity * 1e-3 * length)


def mmHg2Pa(mmHg):
    return mmHg * 133.322
```

# Flow versus radius


```python
import matplotlib.pyplot as plt


conditions = {
    "Normal": {"viscosity": 4, "length": 0.3, "dP": 40},
}

diameter = np.linspace(1e-3, 30e-3, 100)  # Radii from 1mm to 30mm
plt.figure(figsize=(8,5))
for scenario, params in conditions.items():
    # Parameters
    viscosity = params["viscosity"]  # Pa.s (typical blood viscosity)
    length = params["length"]     # m (segment length)
    dP = params["dP"]    #

    # Calculate flow
    Q = poiseuille_flow(dP, diameter / 2, viscosity, length)

    # Plot
    plt.plot(diameter * 1e3, Q*1e3, label=scenario)
    plt.xlabel('Artery diameter (mm)')
    plt.ylabel('Flow Rate ($\\frac{L}{min}$)')
    plt.title("Poiseuille's Law: Flow vs Vessel Diameter")
    plt.grid(True)

plt.legend()
plt.show()
```

```python
import pandas as pd
from pathlib import Path
df0001 = pd.read_csv(Path("files/autonomic-aging-cardiovascular/0001.csv"), delimiter=",")
fs = 100 # samplingrate
bp0001 = df0001["NIBP"].to_numpy()
time = df0001["time_s"].to_numpy()
# Show data df0001
df0001
```

```python
# Beregn det gennemsnitlige
from scipy.signal import find_peaks
peakp_0001 = find_peaks(bp0001, height=bp0001.mean() + 2*bp0001.std()) # du kan finde highpoints, ved bare at bruge det positive signalet
lowp_0001 = find_peaks(-bp0001, height=-bp0001.mean() + 2*bp0001.std()) # du kan finde lowpoints, ved bare at bruge det negative signalet

```


## Arterielle blodtryk


```python
import matplotlib.pyplot as plt
# Plot Continuous Blood Pressures from two subjects
from matplotlib import colors
tableau_colors = list(colors.TABLEAU_COLORS.values())  # ['#1f77b4', '#ff7f0e', ...] det her er bare en list af fine farver vi kan bruge
# Plot Continuous Blood Pressures with peak pressures from two subjects
plt.plot(time, bp0001, label="0001", color=tableau_colors[0])
plt.scatter(time[lowp_0001], bp0001[lowp_0001], color=tableau_colors[0])
plt.scatter(time[peakp_0001], bp0001[peakp_0001], color=tableau_colors[0])

plt.plot(time, bp0002, label="0002", color=tableau_colors[1])

plt.ylabel("BP [mmHg]")
plt.xlabel("time [s]")
plt.legend(loc='upper right')
plt.xlim([0,12])
plt.title(f"Continuous Blood Pressure Over 10 seconds, samplerate={fs}Hz")
plt.show()
```
